# 7. Handle Numerical Missing Values

**Purpose:** Replace missing numerical values using the median.

**Reason:** The median is less affected by extreme values.

```python
for col in numerical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].median()
        )
```

**Purpose:** Replace missing categorical values using the mode.

**Reason:** The mode represents the most frequently occurring category.

```python
for col in categorical_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(
            df_cleaned[col].mode()[0]
        )
```

# 12. Outlier Detection Using Z-Score

**Purpose:** Identify observations that significantly deviate from the mean.

A Z-score greater than 3 or less than -3 is considered a potential outlier.

```python
all_numerical_cols = df.select_dtypes(include='number').columns

z_score_df = pd.DataFrame(index=df.index)

for col in all_numerical_cols:
    mean = df[col].mean()
    std = df[col].std()

    z_score_df[col] = (df[col] - mean) / std

z_outlier_mask = z_score_df.abs() > 3

z_outlier_report = pd.DataFrame({
    "Outlier Count": z_outlier_mask.sum(),
    "Outlier Percentage": (
        z_outlier_mask.sum() / len(df)
    ) * 100
})

z_outlier_report
```

# 12. Outlier Detection Using IQR

**Purpose:** Identify observations that significantly deviate from the middle 50% of the data.

An observation is considered a potential outlier if it is below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR.

```python
# IQR Outlier Detection

import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv(r"C:\Users\ELCOT\Documents\SmartFarming_recommendation.csv")

# Select numerical columns
numerical_cols = df.select_dtypes(include=np.number).columns

# Create outlier mask
iqr_outlier_mask = pd.DataFrame(index=df.index)

# Detect outliers using IQR
for col in numerical_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    iqr_outlier_mask[col] = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    )

# Create outlier report
iqr_outlier_report = pd.DataFrame({
    "Outlier Count": iqr_outlier_mask.sum(),
    "Outlier Percentage":
        (iqr_outlier_mask.sum() / len(df)) * 100
})

# Display report
print("IQR Outlier Report:")
display(iqr_outlier_report)
```

# 1. IMPORT REQUIRED LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder, LabelEncoder, PowerTransformer
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Load the supplied Smart Farming CSV.
# Keep the CSV in the same folder as this notebook.
df = pd.read_csv(r"C:\Users\ELCOT\Downloads\SF24_cleaned (1).csv")

print("Dataset shape:", df.shape)
display(df.head())
```

# 1. Identify numerical, categorical and target columns
Purpose:
Separate predictor columns from the target so that the crop label is never scaled as an input feature.

```python
target_column = "label"
categorical_columns = ["soil_type", "water_source_type", "growth_stage"]
categorical_columns = [c for c in categorical_columns if c in df.columns]
numerical_columns = [
    c for c in df.select_dtypes(include=np.number).columns
    if c not in categorical_columns
]

print("Numerical features:", numerical_columns)
print("Categorical features:", categorical_columns)
print("Target:", target_column)
print("Number of crop classes:", df[target_column].nunique())
```

# 2. Check missing values before feature engineering
Purpose:
Record the missing-value condition and impute missing values before scaling or feature selection.

```python
missing_report = pd.DataFrame({
    "Missing Values": df.isna().sum(),
    "Missing Percentage": (df.isna().mean() * 100).round(2)
})
display(missing_report)

df_work = df.copy()
num_imputer = SimpleImputer(strategy="median")
cat_imputer = SimpleImputer(strategy="most_frequent")

df_work[numerical_columns] = num_imputer.fit_transform(df_work[numerical_columns])
if categorical_columns:
    df_work[categorical_columns] = cat_imputer.fit_transform(df_work[categorical_columns])

print("Missing values after imputation:", int(df_work.isna().sum().sum()))
```

# 3. Analyze numerical distributions before scaling
Purpose: Compare feature ranges and distributions before choosing scaling methods.

```python
display(df_work[numerical_columns].describe().T.round(3))
for col in numerical_columns:
    plt.figure(figsize=(6, 3))
    sns.histplot(df_work[col], kde=True, bins=25)
    plt.title(f"Before Scaling: {col}")
    plt.tight_layout()
    plt.show()
```

# 4. Standardization using StandardScaler
Purpose: Convert numerical features to approximately mean 0 and standard deviation 1.

```python
standard_scaler = StandardScaler()
standard_scaled_array = standard_scaler.fit_transform(df_work[numerical_columns])
df_standard_scaled = pd.DataFrame(
    standard_scaled_array,
    columns=numerical_columns,
    index=df_work.index
)
display(df_standard_scaled.head())
display(df_standard_scaled.describe().T[["mean", "std", "min", "max"]].round(3))
```

# 5. Normalization using MinMaxScaler
Purpose: Transform each numerical feature into the range 0 to 1.

```python
minmax_scaler = MinMaxScaler()
minmax_scaled_array = minmax_scaler.fit_transform(df_work[numerical_columns])
df_minmax_scaled = pd.DataFrame(
    minmax_scaled_array,
    columns=numerical_columns,
    index=df_work.index
)
display(df_minmax_scaled.head())
display(pd.DataFrame({
    "Minimum": df_minmax_scaled.min(),
    "Maximum": df_minmax_scaled.max()
}).round(3))
```

# 6. Compare original, standardized and normalized values
Purpose: To compare the original numerical values with the standardized values produced by StandardScaler and the normalized values produced by MinMaxScaler. This helps understand how both scaling techniques change the range and distribution of numerical features while making them suitable for further analysis.

```python
comparison_columns = numerical_columns[:5]
comparison = pd.concat([
    df_work[comparison_columns].head().add_suffix("_Before"),
    df_standard_scaled[comparison_columns].head().add_suffix("_Standard"),
    df_minmax_scaled[comparison_columns].head().add_suffix("_MinMax")
], axis=1)
display(comparison.round(3))
```

# 7. Encode categorical predictor variables
Purpose: Convert categorical information into numeric columns for machine learning.

```python
df_encoded = df_work.copy()
df_encoded = pd.get_dummies(
    df_encoded,
    columns=categorical_columns,
    dtype=int
)

label_encoder = LabelEncoder()
df_encoded["label_encoded"] = label_encoder.fit_transform(df_encoded[target_column])

print("Crop label mapping:")
display(pd.DataFrame({
    "Crop": label_encoder.classes_,
    "Encoded Value": range(len(label_encoder.classes_))
}))
print("Encoded dataset shape:", df_encoded.shape)
display(df_encoded.head())
```

# 8. Skewness analysis
Purpose:Skewness measures the asymmetry of a feature distribution. Absolute skewness greater than 1 is treated as a candidate for transformation. The target and categorical codes are excluded.

```python
skewness_before = df_work[numerical_columns].skew()
skew_report = pd.DataFrame({
    "Feature": skewness_before.index,
    "Skewness Before": skewness_before.values
})
skew_report["Interpretation"] = skew_report["Skewness Before"].apply(
    lambda x: "Highly positively skewed" if x > 1
    else ("Highly negatively skewed" if x < -1
          else "Approximately symmetric / moderately skewed")
)
display(skew_report.sort_values("Skewness Before", key=lambda s: s.abs(), ascending=False).round(3))
```

# 9. Apply Yeo-Johnson transformation only to highly skewed features
Purpose: Reduce strong skewness where appropriate.

```python
features_to_transform = [
    col for col in numerical_columns
    if abs(skewness_before[col]) > 1 and df_work[col].nunique() > 2
]

df_transformed = df_work.copy()
if features_to_transform:
    power_transformer = PowerTransformer(method="yeo-johnson", standardize=False)
    df_transformed[features_to_transform] = power_transformer.fit_transform(
        df_work[features_to_transform]
    )

skewness_after = df_transformed[numerical_columns].skew()
skew_comparison = pd.DataFrame({
    "Feature": numerical_columns,
    "Skewness Before": [skewness_before[c] for c in numerical_columns],
    "Skewness After": [skewness_after[c] for c in numerical_columns]
})
skew_comparison["Absolute Skewness Reduction"] = (
    skew_comparison["Skewness Before"].abs() -
    skew_comparison["Skewness After"].abs()
)
display(skew_comparison.sort_values(
    "Absolute Skewness Reduction", ascending=False
).round(3))
print("Features transformed:", features_to_transform)
```

# 10. Visual comparison of distributions after transformation
Purpose: To visually compare the distribution of numerical features before and after applying skewness correction transformations using histograms and KDE plots. This helps identify changes in data distribution and check whether the transformation has reduced skewness and made the data more suitable for further analysis.

```python
for col in features_to_transform:
    fig, axes = plt.subplots(1, 2, figsize=(10, 3))
    sns.histplot(df_work[col], kde=True, bins=25, ax=axes[0])
    axes[0].set_title(f"Before: {col}")
    sns.histplot(df_transformed[col], kde=True, bins=25, ax=axes[1])
    axes[1].set_title(f"After Yeo-Johnson: {col}")
    plt.tight_layout()
    plt.show()
```

4.soil_type, water_source_type, and growth_stage are treated as categorical codes and one-hot encoded.

5.The crop target is encoded separately and is not scaled.

6.Yeo-Johnson transformation is applied only to numerical features with absolute skewness greater than 1. The before/after skewness table and histograms should be used to judge whether skewness actually decreased.

Conclusion: Week 6 prepares the Smart Farming data for feature selection and PCA. The original dataset is preserved, and separate imputed, encoded, scaled and transformed outputs are saved.

# WEEK 7: FEATURE SELECTION AND DIMENSIONALITY REDUCTION
Objective:
Identify useful predictor features, reduce redundant information, and apply Principal Component Analysis (PCA). This follows the rule book's Week 7 milestone: feature selection, PCA implementation, and visualization of explained variance.

```python
from sklearn.feature_selection import VarianceThreshold, SelectKBest, f_classif
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer

# Use Week 6 encoded predictors. Exclude original text target and encoded target from X.
feature_df = df_encoded.copy()
X = feature_df.drop(columns=["label", "label_encoded"])
y = feature_df["label_encoded"].copy()

X = X.apply(pd.to_numeric, errors="coerce")
X = X.replace([np.inf, -np.inf], np.nan)
X = X.dropna(axis=1, how="all")
X = pd.DataFrame(
    SimpleImputer(strategy="median").fit_transform(X),
    columns=X.columns,
    index=X.index
)

print("Rows:", X.shape[0])
print("Input feature count:", X.shape[1])
print("Target classes:", y.nunique())
```

# 1. Variance-based feature selection
Purpose: Remove features that have almost no variation. Such features provide little information for distinguishing observations.

```python
variance_selector = VarianceThreshold(threshold=0.01)
X_variance_array = variance_selector.fit_transform(X)
variance_features = X.columns[variance_selector.get_support()].tolist()
X_variance = pd.DataFrame(
    X_variance_array,
    columns=variance_features,
    index=X.index
)

variance_report = pd.DataFrame({
    "Feature": X.columns,
    "Variance": X.var().values,
    "Selected": variance_selector.get_support()
}).sort_values("Variance")
display(variance_report)
print("Features before:", X.shape[1])
print("Features after variance filtering:", X_variance.shape[1])
```

# 2. Correlation-based feature analysis
Purpose: Identify highly correlated predictor pairs that may contain overlapping information.

```python
corr_matrix = X_variance.corr()
plt.figure(figsize=(14, 10))
sns.heatmap(corr_matrix, cmap="coolwarm", center=0)
plt.title("Correlation Matrix after Variance Filtering")
plt.tight_layout()
plt.show()

upper_triangle = corr_matrix.abs().where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)
correlation_drop = [
    column for column in upper_triangle.columns
    if (upper_triangle[column] > 0.90).any()
]
X_corr = X_variance.drop(columns=correlation_drop)

print("Highly correlated features removed:", correlation_drop)
print("Features after correlation filtering:", X_corr.shape[1])
```

# 3. Statistical feature selection using SelectKBest
Purpose: Rank features according to their statistical relationship with the crop target.

```python
k = min(10, X_corr.shape[1])
selector = SelectKBest(score_func=f_classif, k=k)
X_selected_array = selector.fit_transform(X_corr, y)
selected_features = X_corr.columns[selector.get_support()].tolist()

feature_scores = pd.DataFrame({
    "Feature": X_corr.columns,
    "F Score": selector.scores_,
    "P Value": selector.pvalues_,
    "Selected": selector.get_support()
}).sort_values("F Score", ascending=False)

display(feature_scores.round(5))
X_selected = pd.DataFrame(
    X_selected_array,
    columns=selected_features,
    index=X.index
)
print("Selected features:", selected_features)
```

# 4. Visualize selected feature importance
Purpose: To visualize the importance scores of selected features using a bar chart. This helps identify which features contribute most to the target variable and which features are less important, making it easier to select relevant features for further analysis and PCA.

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=feature_scores.head(k),
    x="F Score",
    y="Feature"
)
plt.title("Top Features Selected by SelectKBest")
plt.tight_layout()
plt.show()
```

# 5. Principal Component Analysis (PCA)
Purpose: Convert the selected correlated features into a smaller set of principal components while retaining as much variance as possible.

```python
pca_scaler = StandardScaler()
X_selected_scaled = pca_scaler.fit_transform(X_selected)

pca = PCA()
X_pca_array = pca.fit_transform(X_selected_scaled)

explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

variance_table = pd.DataFrame({
    "Principal Component": [f"PC{i+1}" for i in range(len(explained_variance))],
    "Explained Variance Ratio": explained_variance,
    "Cumulative Explained Variance": cumulative_variance
})
display(variance_table.round(4))
```

# 6. Explained variance visualization
Purpose:The first chart shows the variance explained by each component. The second shows cumulative explained variance, which helps decide how many components are needed to retain a chosen proportion of information.

```python
plt.figure(figsize=(8, 4))
plt.bar(range(1, len(explained_variance) + 1), explained_variance)
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("PCA Explained Variance Ratio")
plt.xticks(range(1, len(explained_variance) + 1))
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o"
)
plt.axhline(y=0.95, linestyle="--", label="95% variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Cumulative Explained Variance")
plt.xticks(range(1, len(cumulative_variance) + 1))
plt.legend()
plt.tight_layout()
plt.show()

n_components_95 = int(np.argmax(cumulative_variance >= 0.95) + 1)
print("Components required to retain at least 95% variance:", n_components_95)
```

# 7. 2D PCA scatter plot
Purpose:The scatter plot displays observations using the first two principal components. Points are colored by crop label to inspect whether crop classes form visible groups. Overlap is possible and should not be interpreted as a classification result.

```python
if X_pca_array.shape[1] >= 2:
    pca_2d = pd.DataFrame({
        "PC1": X_pca_array[:, 0],
        "PC2": X_pca_array[:, 1],
        "Crop": df_work[target_column].values
    })
    plt.figure(figsize=(10, 7))
    sns.scatterplot(
        data=pca_2d,
        x="PC1",
        y="PC2",
        hue="Crop",
        s=25,
        alpha=0.7,
        legend="full"
    )
    plt.title("Smart Farming Dataset: PCA 2D Projection")
    plt.tight_layout()
    plt.show()
```

# 8. Create and save the reduced datasets
Purpose: To create and save reduced datasets containing only the selected important features and PCA-transformed components. This helps reduce the number of features, simplify the dataset, remove unnecessary information, and prepare the data for further analysis and visualization.

```python
pca_95 = PCA(n_components=n_components_95)
X_pca_95 = pca_95.fit_transform(X_selected_scaled)

df_feature_selected = X_selected.copy()
df_feature_selected["label"] = df_work[target_column].values

pca_columns = [f"PC{i+1}" for i in range(X_pca_95.shape[1])]
df_pca_reduced = pd.DataFrame(X_pca_95, columns=pca_columns, index=X.index)
df_pca_reduced["label"] = df_work[target_column].values

df_feature_selected.to_csv("smart_farming_week7_selected_features.csv", index=False)
df_pca_reduced.to_csv("smart_farming_week7_pca_reduced.csv", index=False)
feature_scores.to_csv("smart_farming_week7_feature_scores.csv", index=False)
variance_table.to_csv("smart_farming_week7_pca_variance.csv", index=False)

print("Feature-selected dataset shape:", df_feature_selected.shape)
print("PCA-reduced dataset shape:", df_pca_reduced.shape)
print("Week 7 output files saved successfully.")
```
